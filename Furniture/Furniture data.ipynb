**Dataset Information:**

Each column provides specific information about the order, the customer, and the product sold, making this dataset suitable for various sales and profitability analysis tasks in the retail domain. Here's a brief explanation of each column in the dataset -

**Order ID:** This column represents the unique identifier of the customer's order.

**Order Date:** This column represents the date on which the customer placed the order.

**Ship Date:** This column represents the date on which the order was shipped.

**Ship Mode:** This column represents the shipping method used to deliver the order (e.g., Standard, First, Second, Same Day).

**Customer ID:** This column represents the unique identifier assigned to each customer.

**Customer Name:** This column represents the name of the customer who placed the order.

**Segment:** This column represents the customer segment the order belongs to (e.g., Consumer, Corporate, Home Office).

**Country:** This column represents the country where the order was placed.

**City:** This column represents the city where the order was placed.

**State:** This column represents the state where the order was placed.

**Region:** This column represents the geographic region the order belongs to (e.g., East, West, Central, South).

**Product ID:** This column represents the unique identifier assigned to each product.

**Category:** This column represents the main product category (e.g., Furniture).

**Sub-Category:** This column represents the specific product sub-category (e.g., Chairs, Tables, Bookcases).

**Product Name:** This column represents the name of the product that was ordered.

**Sales:** This column represents the sales amount generated by the product line in the order, in dollars.

**Quantity:** This column represents the number of units sold.

**Profit:** This column represents the profit or loss generated by the order, in dollars.

**Duration:** This column represents the number of days between the order date and the ship date.

**Month:** This column represents the month in which the order was placed.

### <b><span style='color:#DEB078'>5.1 |</span><span style='color:#E13213'> Numerical Features</span></b> 

```python
# KPI
Revenues = df['Sales'].sum()
Profit = df['Profit'].sum()
Total_Order = df['Order ID'].nunique()
Quantity = df['Quantity'].sum()
Profit_Margin = round(Profit / Revenues * 100, 2)
Avg_Order_Value = round( Revenues / Total_Order, 2) 
```

```python
kpi = pd.DataFrame({
    'Metric': ['Revenues','Profit','Total_Order','Quantity', 'Profit_Margin','Avg_Order_Value'],
    'Value': [f"${Revenues:,.2f}",f"${Profit:,.2f}", Total_Order, Quantity, f"{Profit_Margin}%", f"${Avg_Order_Value:,.2f}"]
})

kpi
```

### <b><span style='color:#DEB078'>5.2 |</span><span style='color:#E13213'> Categorical Features</span></b> 

```python
df.describe(include=['O'])
```

```python
cols = ['Ship Mode', 'Segment', 'Region', 'Sub-Category', 'Duration']
```

```python
for i in cols:
    fig, ax = plt.subplots(1, 2, figsize=(10,4))
    fig.suptitle(i, fontsize=16)
    
    df[i].value_counts().plot(kind='bar', ax=ax[0])
    df[i].value_counts().plot(kind='pie', autopct='%.1f%%', ax=ax[1])
    
    plt.show()
```

```python
# Which sales territory yields the highest performance?
Region = df.groupby("Region")["Sales"].sum()
plt.bar(Region.index, Region.values)
plt.title("Region By Sales")
plt.xlabel("Region")
plt.ylabel("Sales")
plt.show()
```

```python
# Which product sub-category generates the most sales?
Sub_Category = df.groupby("Sub-Category")["Sales"].sum()

Sub_Category = Sub_Category.sort_values(ascending=True)

print("Sub-Category:", Sub_Category.idxmax())
print("Highest Sales:", Sub_Category.max())

plt.barh(Sub_Category.index, Sub_Category.values)

plt.title("Sub-Category By Sales")
plt.xlabel("Sales")
plt.ylabel("Sub-Category")
plt.show()
```

```python
# Which months have the highest sales?
monthly_sales = df.groupby('Month')['Sales'].sum()
month_order = ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
monthly_sales = monthly_sales.reindex(month_order)

plt.grid()
plt.title("Total Sales By Month (2014-2017)")
plt.plot(monthly_sales.index, monthly_sales.values, marker='o')
plt.show()
```

```python
# Which state-segment combinations bring in the most sales?
State_segment = (df.groupby(["State","Segment"])["Sales"].sum()
                   .sort_values(ascending=False).head(10).reset_index())

ax= sns.barplot(data = State_segment, x= "Sales", y= "State", hue = "Segment")

for container in ax.containers:
        ax.bar_label(container, fmt='{:,.0f}', padding=3)
    
ax.margins(x=0.2)
plt.title("Top 10 State-Segment Combinations By Sales")
plt.show()
```

```python
# Which ship mode delivers the fastest?
df["Duration_days"] = df["Duration"].str.extract(r'(\d+)').astype(int)
avg_duration =(df.groupby("Ship Mode")["Duration_days"].mean()
                        .sort_values(ascending=True).round(2).reset_index())

ax= sns.barplot(data= avg_duration, x= "Ship Mode", y= "Duration_days")
ax.bar_label(ax.containers[0], fmt='%.2f')

plt.title("Average Delivery Duration By Ship Mode")
plt.show()
```

```python
# Which regions are profitable?
Region_Profit = df.groupby("Region")["Profit"].sum().sort_values()

colors = ["#c1121f" if v < 0 else "#2E86AB" for v in Region_Profit.values]
bars = plt.bar(Region_Profit.index, Region_Profit.values, color=colors)
plt.bar_label(bars, fmt='{:,.0f}', padding=3)

plt.axhline(0, color="black", linewidth=0.8)
plt.margins(y=0.15)
plt.title("Region By Profit")
plt.xlabel("Region")
plt.ylabel("Profit")
plt.show()
```
